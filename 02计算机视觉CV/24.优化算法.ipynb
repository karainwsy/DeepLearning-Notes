##### ***优化问题***
###### 优化问题的一般形式是：限定条件下目标函数要达到最大或者最小的问题。且分为局部优化问题和全局优化问题。其中使用迭代优化算法来求解，一般只能保证找到局部最小值，而不能保证找到全局最小值。
###### 凸优化问题是指目标函数和约束条件都是凸函数的优化问题。凸函数的性质是：函数的任意两点之间的连线都在函数的曲线上方。对于凸优化问题，如果得到的最优解是局部最优解，那么它也是全局最优解。严格的凸优化问题有位移的全局最优解。
###### 遗憾的是，很多实际问题的目标函数和约束条件都不是凸函数，因此不能使用凸优化算法来求解。目前学到的凸函数有线性回归，softmax回归。而如MLP，CNN，RNN，attention等模型的目标函数和约束条件都不是凸函数，因此不能使用凸优化算法来求解。从直觉上来理解，因为凸函数的表达能力很弱，所以对于复杂的问题往往处理的不是很好，而观察函数结构，线性回归和softmax都没有相应的非线性函数激活块，所以会导致函数的表达能力不足。而诸如MLP，RNN，CNN等都有着激活函数为其引入了非线性性。
###### 然而即使在深度学习中的优化问题通常都是非凸的，但他们也经常在局部极小值附近表现出一些凸性，所以在局部极小值附近使用凸优化算法往往能够得到比较好的结果。
##### ***优化算法：梯度下降***
###### 尽管梯度下降很少直接应用于深度学习，但了解它的理解随机梯度下降算法的关键。梯度下降算法是当前样本点减去梯度的方向，乘以一个学习率，得到下一个样本点。但是问题随之而来，当样本比较多的时候，梯度下降算法的计算量就会比较大，因为每次都需要计算所有样本的梯度。再深度学习中往往f(x)是损失函数的平均，这样会导致大量的计算，成本过高，所以在深度学习中常用它的变种：随机梯度下降。
##### ***随机梯度下降***
###### 随机梯度下降算法是每次随机选择一个样本，计算它的梯度，然后更新参数。这样每次只需要计算一个样本的梯度，成本大大降低。但是由于每次只计算一个样本的梯度，所以不能保证每次迭代都能朝着全局最优解的方向走。而且单样本的梯度下降难以完全利用硬件资源，所以产生了小批量随机梯度下降。

##### ***小批量随机梯度下降***
###### 小批量随机梯度下降算法是每次随机选择一个小批量的样本，计算它的梯度，然后更新参数。这样每次只需要计算一个小批量样本的梯度，成本大大降低。而且可以充分利用硬件资源，并行计算。

##### ***冲量法***
###### 虽然小批量随机梯度下降最为加速计算的手段，但它也有一些问题。比如在训练过程中，可能会出现震荡的情况，即参数在最优解附近波动，而不是收敛到最优解。为了解决这个问题，提出了冲量法。
###### 冲量法是在小批量随机梯度下降的基础上，引入了一个动量项。动量项是上一次迭代的梯度方向，乘以一个动量系数，得到当前迭代的动量项。然后将当前迭代的动量项加到当前迭代的梯度方向上，得到当前迭代的梯度。这样可以在震荡的情况下，保持梯度的方向能更加稳定，从而加速收敛。

###### 其中的动量系数β一般取[0.5, 0.9, 0.95, 0.99]。
##### ***Adam算法***
###### 上述的算法出现震荡的原因大都因为学习率的设置不当，导致会在最优解附近震荡而难以收敛，而Adam算法是一种自适应学习率的优化算法，它可以根据梯度的变化动态调整学习率，从而避免震荡问题。
