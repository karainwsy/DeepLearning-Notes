##### ***循环神经网络（Recurrent Neural Network, RNN）***
###### 根据N元语法我们可知，为增强模型的性能，我们需要增加n，即增加考虑的词元数量。然而随之带来的问题是模型参数的数量会随之呈指数增长。所以与其将$P(x_t|x_{t-1},\cdots,x_{t-n+1})$模型化，不如使用隐变量模型将其看做一个隐状态或称之为隐藏变量。$P(x_t|x_{t-1},\cdots,x_1)\approx P(x_t|h_{t-1})$<br>其中隐状态可以通过当前输入$x_t$和先前隐状态$h_{t-1}$来计算时间步$t$处的任何时间的隐状态：$h_t = f(x_t, h_{t-1})$。<br>不同于隐藏层，隐状态是个截然不同的概念。前者无时间概念，只对当前层有效，且每层独立参数，只进行特征提取；后者则有时间维度，依赖上一时刻的隐状态，一般承担记忆功能。
###### 由此得出的公式为：$H_t = \sigma(X_T W_{xh} + H_{t-1} W_{hh} + b_h)$与普通的隐藏层不同，这项公式中多了一项$H_{t-1}W_{hh}$，它依赖于上一时刻的隐状态$h_{t-1}$，所以RNN具有记忆功能，该层也叫做循环层。

###### 其中循环神经网络模型的参数数量不会随着时间步的增加而增加，也就是说RNN的参数在一开始就固定且后续一直循环使用。而普通的MLP模型的参数数量会随着时间步的增加而增加，因为每层都有自己的参数。
##### ***困惑度（Perplexity）***
###### 为衡量语言模型的性能我们引入了一个指标叫困惑度（Perplexity），该指标通过计算序列的似然概率来度量模型的质量，然而这是一个难以理解，难以比较的数字。不同于概率是描述已知参数时的随机变量的输出结果，似然则用来描述已知随机变量输出结果，参数的可能取值。毕竟较短的序列出现的概率比较长的序列更高，若是简单的以概率来衡量会导致更短的序列出现概率更高从而导致不公平，所以使用似然来进行预测。然而由于概率会越乘越小，所以我们可以通过去对数把乘法变成加法，再取个负号就可以得到一个loss函数:
$$\frac{1}{n}\sum_{t=1}^n-\log P(x_t|x_{t-1},\cdots,x_1)$$
###### 为了能更直观的理解困惑度，我们可以将其取指数，得到一个困惑度值。困惑度值越小，模型的性能越好。即$$Perplexity = \exp(-\frac{1}{n}\sum_{t=1}^n\log P(x_t|x_{t-1},\cdots,x_1))$$
