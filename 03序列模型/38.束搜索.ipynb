##### ***搜索策略***
###### 在之前的小节中，通过给定训练数据，从而训练出一个模型来预测下一个词的概率分布，而之后如何选取输出的词就要采用搜索策略。
##### ***贪心搜索***
###### 贪心搜索是一种简单的搜索策略，它在每个时间步都选择概率最大的词作为下一个词。然而这个策略的问题在于，现实中的最优序列应该是最大化$\prod_{t'=1}^{T'}P(y_{t'}|y_{1},\cdots,y_{t'-1},c)$值的输出序列，这是基于输入序列生成输出序列的条件概率，然而贪心搜索得到的结果无法保证是全局最优解。
##### ***穷举搜索***
###### 而如果从贪心搜索的问题出发，想解决它的办法就很显而易见了，利用穷举法将所有可能的输出序列及其概率计算出来，然后选择概率最大的序列作为最优解。然而它的问题非常致命，因为输出序列的数量是指数级的，所以计算成本非常高。只有在序列长度非常短时才有可能考虑，不过一般不考虑。
##### ***束搜索***
###### 知道了上述两种搜索策略的问题后，我们就可以选择一种介于两者之间的搜索策略。束搜索是贪心搜索的一个改进版本，它引入了一个超参数：束宽（beam size）$k$。在时间步1，我们选择概率最大的$k$个词元作为输出序列的第一个词元，在随后的每个时间步，基于上一个时间步的k个候选序列，继续选择其中具有最高条件概率的k个候选输出序列，由此得到候选输出序列集合，而后选择其中条件概率乘积最高的序列作为输出序列：
$$\frac{1}{L^{\alpha}}logP(y_1, \cdots, y_L|c) = \frac{1}{L^{\alpha}}\sum_{t'=1}^LlogP(y_{t'}|y_{1},\cdots,y_{t'-1},c)$$
###### 式中，L是最终候选序列的长度，$\alpha$通常设置成0.75.因为一个较长的序列在求和中会有更多的对数项，因为分母中的$L^{\alpha}$用于惩罚长序列。
