```python
# 1.梯度消失与梯度爆炸
# 在复杂的神经网络中由于反向传播要进行链式法则，所以链式法则上的每一项的梯度如果很大，就会导致很多大于1的数相乘，导致梯度爆炸。反之如果梯度很小，就会导致很多小于1的数相乘，导致梯度消失。

# 梯度爆炸的问题：值会超出值域范围变成inf，其次对学习率会很敏感
# 梯度消失的问题：值会趋近于0，对学习率也会很敏感
```

```python
# 解决方法：
# 1.将梯度控制在一定数值范围内
# 2.将乘法转换为加法
# 3.引入正则化项，归一化梯度数值或者对梯度进行裁剪
# 4.合理的初始化参数和激活函数与学习率
```

```python
# 参数初始化：假设权重是一个独立的同分布并且假设没有激活函数
# 让每层的方差是一个常数，也就是说将每层的输出和梯度看做随机变量，让它们的均值和方差在每一层都保持一致。每一层的均值为0，方差为1。
# 为了达到这个目标，我们初始化均值为零，可以证明无论后续的层有多少，它们的均值都保持为零。
# 而它们的方差我们希望初始化是一个常数，通过数学计算为了保证后续层的方差保持一致，我们需要将权重的方差初始化为1/上一层的神经元数量。而在反向计算中，我们需要将梯度的方差初始化为1/下一层的神经元数量。
# 以上就是著名的Xavier初始化方法。
```

```python
# 不过要想同时满足这个两个条件很困难，一般只有在输入和输出的神经元数量一致时才会满足。所以选取折中情况，将权重的方差初始化为2/（上一层的神经元数量+下一层的神经元数量）。
```

```python
# 对于线性激活函数我们要满足上述的条件，我们需要使这个激活函数为y=αx+β，其中α=1，β=0。检查常用的激活函数发现，tanh和ReLu函数在原点附近都满足这个条件。而Sigmoid函数需要乘上一个系数加上一个偏移量，才能满足这个条件。
```
